# Задание 1

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv(
    "school_progress.csv",
    index_col="enrollment_id"
)


average_scores = data.groupby("grade")["average_score"].mean()

plt.plot(
    average_scores.index,
    average_scores.values,
    marker="o"
)

plt.title("Средний average_score для каждого grade")
plt.xlabel("Grade")
plt.ylabel("Average score")
plt.grid(True)

plt.show()
```

# Задание 2

```python
course_counts = data["course"].value_counts()

# Строим горизонтальную столбчатую диаграмму
plt.figure(figsize=(10, 6))
course_counts.sort_values().plot(kind="barh")

plt.title("Популярность курсов")
plt.xlabel("Количество записей")
plt.ylabel("Название курса")
plt.grid(axis="x")

plt.tight_layout()
plt.show()
```

```python


fig, axes = plt.subplots(1, 3, figsize=(15, 5))

bins_values = [5, 10, 20]

for ax, bins in zip(axes, bins_values):
    ax.hist(data["average_score"].dropna(), bins=bins, edgecolor="black")
    ax.set_title(f"Количество интервалов: {bins}")
    ax.set_xlabel("Средний балл")
    ax.set_ylabel("Количество учеников")
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.scatter(
    data["hours_spent"],
    data["average_score"],
     alpha=0.5
)

plt.title("Зависимость среднего балла от времени занятий")
plt.xlabel("Время занятий (часы)")
plt.ylabel("Средний балл")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
data["status"].value_counts().plot(
    kind="bar",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Количество записей по статусам")
axes[0, 0].set_xlabel("Статус")
axes[0, 0].set_ylabel("Количество")


axes[0, 1].hist(data["average_score"].dropna(), bins=10, edgecolor="black")
axes[0, 1].set_title("Распределение среднего балла")
axes[0, 1].set_xlabel("Средний балл")
axes[0, 1].set_ylabel("Количество записей")


axes[1, 0].scatter(
    data["hours_spent"],
    data["average_score"],
    alpha=0.5
)
axes[1, 0].set_title("Время занятий и средний балл")
axes[1, 0].set_xlabel("Время занятий (часы)")
axes[1, 0].set_ylabel("Средний балл")


average_scores = data.groupby("grade")["average_score"].mean()

axes[1, 1].plot(
    average_scores.index,
    average_scores.values,
    marker="o"
)
axes[1, 1].set_title("Средний балл по классам")
axes[1, 1].set_xlabel("Класс")
axes[1, 1].set_ylabel("Средний балл")


plt.tight_layout()
plt.show()
```
